```python
import numpy as np

x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])

values, counts = np.unique(x, return_counts=True)
max_count = counts.max()
most_frequent = values[counts == max_count]

print(most_frequent)
print(np.where(x == most_frequent[0])[0])
```

```python
import numpy as np

y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

values, counts = np.unique(y, return_counts=True)
max_count = counts.max()
most_frequent = values[counts == max_count]

print(most_frequent)

for value in most_frequent:
    print(np.where(y == value)[0])
```

```python
import numpy as np

gfg = np.matrix('[4, 1, 9; 12, 3, 1; 4, 5, 6]')

print(gfg.sum())
print(gfg.sum(axis=1))
print(gfg.sum(axis=0))
```

```python
import numpy as np

n_array = np.array([[55, 25, 15], [30, 44, 2], [11, 45, 77]])

print(np.trace(n_array))

eigenvalues, eigenvectors = np.linalg.eig(n_array)

print(eigenvalues)
print(eigenvectors)

print(np.linalg.inv(n_array))

print(np.linalg.det(n_array))
```

```python
import numpy as np

p = np.array([[1, 2], [2, 3]])
q = np.array([[4, 5], [6, 7]])

print(p @ q)
print(np.cov(p.flatten(), q.flatten()))
```

```python
import numpy as np

p = np.array([[1, 2], [2, 3], [4, 5]])
q = np.array([[4, 5, 1], [6, 7, 2]])

print(p @ q)
print(np.cov(p.flatten(), q.flatten()))
```

```python
import numpy as np
from itertools import product

x = np.array([[2, 3, 4], [3, 2, 9]])
y = np.array([[1, 5, 0], [5, 10, 3]])

print(np.inner(x, y))
print(np.outer(x, y))
print(list(product(x.flatten(), y.flatten())))
```

```python
import numpy as np

array = np.array([[1, -2, 3], [-4, 5, -6]])

print(np.abs(array))

print(np.percentile(array, [25, 50, 75]))
print(np.percentile(array, [25, 50, 75], axis=0))
print(np.percentile(array, [25, 50, 75], axis=1))

print(np.mean(array))
print(np.median(array))
print(np.std(array))

print(np.mean(array, axis=0))
print(np.median(array, axis=0))
print(np.std(array, axis=0))

print(np.mean(array, axis=1))
print(np.median(array, axis=1))
print(np.std(array, axis=1))
```

```python
import numpy as np

array = np.array([1.0, 1.2, 2.2, 2.0, 3.0, 2.0])

integer_elements = array[array == array.astype(int)]
float_elements = array[array != array.astype(int)]

print(integer_elements)
print(float_elements)
```

```python
from PIL import Image
import numpy as np

def img_to_array(path):
    image = Image.open(path)
    img_array = np.array(image)

    if len(img_array.shape) == 3:
        np.savetxt("image_rgb.txt", img_array.reshape(-1, 3), fmt='%d')
    elif len(img_array.shape) == 2:
        np.savetxt("image_grayscale.txt", img_array, fmt='%d')

img_to_array("WIN_20260908_12_15_09_Pro.jpg")
```
