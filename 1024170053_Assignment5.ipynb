```python
import numpy as np

x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])

values, counts = np.unique(x, return_counts=True)

most_frequent = values[np.argmax(counts)]
indices = np.where(x == most_frequent)[0]

print("Most frequent value:", most_frequent)
print("Indices:", indices)
```

```python
import numpy as np

y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

values, counts = np.unique(y, return_counts=True)

most_frequent = values[np.argmax(counts)]
indices = np.where(y == most_frequent)[0]

print("Most frequent value:", most_frequent)
print("Indices:", indices)
```

```python
import numpy as np

DeepanjanBaral = np.linspace(10, 100, 25)

print(DeepanjanBaral)
print("Dimensions:", DeepanjanBaral.ndim)
print("Shape:", DeepanjanBaral.shape)
print("Total elements:", DeepanjanBaral.size)
print("Data type:", DeepanjanBaral.dtype)
print("Total bytes:", DeepanjanBaral.nbytes)

transpose = DeepanjanBaral.reshape(25, 1)

print("Transpose:")
print(transpose)

print("Using T attribute:")
print(DeepanjanBaral.T)
```
